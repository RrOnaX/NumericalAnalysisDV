```python
#Задача 1.
#Пункт 1.

import matplotlib.pyplot as plt

def f(x):
    return x**3 - 6 * x**2 + 11 * x - 6

x = []
y = []
step = 0.0
while step <= 4.0:
    x.append(step)
    y.append(f(step))
    step += 0.05

plt.plot(x, y)
plt.axhline(0, color="black")
plt.grid(True)
plt.savefig("graph.png")
plt.show()
```

```python
#Пункт 2.

def f(x):
    return x**3 - 6 * x**2 + 11 * x - 6

def bisection(a, b, eps=1e-9):
    k = 0
    while b - a >= eps:
        c = (a + b) / 2
        k += 1
        if f(c) == 0:
            break
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2, k

print(bisection(0.4, 1.5))
print(bisection(1.4, 2.6))
print(bisection(2.2, 3.7))
```

```python
def f(x):
    return x**3 - 6 * x**2 + 11 * x - 6

def bisection(a, b, eps=1e-9):
    journal = []
    k = 0
    while b - a >= eps:
        c = (a + b) / 2
        k += 1
        journal.append([k, round(a, 6), round(b, 6), round(c, 6), round(f(c), 9), round(b - a, 9)])
        if f(c) == 0:
            break
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2, journal

for a, b in [(0.4, 1.5), (1.4, 2.7), (2.2, 3.7)]:
    root, j = bisection(a, b)
    print(f"\nИнтервал [{a}, {b}] -> корень: {root}")
    print("k | a | b | c | f(c) | b-a")
    for row in j:
        print(row)
```

```python
#Пункт 4.

import math

eps = 1e-9
intervals = [(0.4, 1.5), (1.4, 2.7), (2.2, 3.7)]

for a, b in intervals:
    root, j = bisection(a, b, eps)
    d0 = b - a
    teor = math.ceil(math.log2(d0 / eps))
    print(f"[{a}, {b}] | длина = {round(d0, 2)} | итераций = {len(j)} | по формуле = {teor}")
```

```python
#Пункт 5.

tasks = [
    (0.4, 1.5, 1.0),
    (1.4, 2.7, 2.0),
    (2.2, 3.7, 3.0)
]

print("Точный | Найденный | Итераций | Δ (абс.) | δ (отн.)")
print("-" * 60)

for a, b, exact in tasks:
    root, j = bisection(a, b, 1e-9)
    delta_abs = abs(root - exact)
    delta_rel = delta_abs / abs(exact)
    print(f"{exact} | {round(root, 10)} | {len(j)} | {delta_abs:.2e} | {delta_rel:.2e}")
```
