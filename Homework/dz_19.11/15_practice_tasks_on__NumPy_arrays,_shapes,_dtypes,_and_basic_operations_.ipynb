```python
#Задача 1. Блок 1
import numpy as np

arr = np.arange(60).reshape(3, 4, 5)

print("shape:", arr.shape)
print("ndim:", arr.ndim)
print("size:", arr.size)
print("dtype:", arr.dtype)
print("strides:", arr.strides)
print("C_CONTIGUOUS:", arr.flags["C_CONTIGUOUS"])
```

В strides = (160, 40, 8) числа показывают сдвиг в байтах при шаге на 1 по осям 160 к следующему блоку 4*5, 40 - к следующей строке из 5 чисел, 8 — к следующему элементу (int64 = 8 байт). size = 60 - это общее число элементов то есть 3 * 4 * 5 = 60

```python
#Задача 2. Блок 1

import numpy as np

a = np.array([1, 2, 3, 4])

print(a / 2)
print(a // 2)
print(a / 2.0)

print(a.dtype)
print((a / 2).dtype)
print((a // 2).dtype)

b = np.array([1, 2, 3], dtype=np.int8)
c = b + 127

print(c)
print(c.dtype)

```

1. При делении создаётся новый массив с дробными числами (float64), а исходный массив a остаётся нетронутым
2. Тип int8 вмещает числа только до 127, поэтому при превышении лимита происходит переполнение и счётчик перескакивает в минус (-128, -127, -126)
3. NumPy не проверяет выход за границы ради максимальной скорости вычислений


```python
#Задача 3. Блок 1

import numpy as np

err32 = float(np.float32(0.1)) - 0.1
err64 = float(np.float64(0.1)) - 0.1

print(repr(err32))
print(f"{err32:.20e}")

print(repr(err64))
print(f"{err64:.20e}")

ratio = np.finfo(np.float32).eps / np.finfo(np.float64).eps
print(ratio)
```

Ошибка float32 превышает ошибку float64 примерно в 536 миллионов раз, поэтому float32 допустим только там, где важнее скорость и экономия памяти при работе с большими массивами (например, в нейросетях и 3D-графике), но категорически не подходит для точных инженерных, научных и финансовых расчётов, где накапливаются погрешности

```python
#Задача 4. Блок 1

import numpy as np

a1 = np.zeros((2, 3), dtype=np.float64)
a2 = np.zeros((2, 3), dtype=np.float32)
a3 = np.ones((2, 3), dtype=np.int16)
a4 = np.full((2, 3), 7, dtype=np.uint8)

print(a1.dtype, a1.nbytes)
print(a2.dtype, a2.nbytes)
print(a3.dtype, a3.nbytes)
print(a4.dtype, a4.nbytes)

big64 = np.zeros((1000, 1000), dtype=np.float64)
big32 = np.zeros((1000, 1000), dtype=np.float32)
print(big64.nbytes - big32.nbytes)
```

Использование float32 вместо float64 экономит ровно половину памяти (4 000 000 байт на массиве 1000x1000) и ускоряет расчёты ценой снижения точности вычислений, поэтому более "лёгкие" типы данных стоит выбирать для больших массивов, где допустима небольшая погрешность.

```python
#Задача 5. Блок 2

import numpy as np

a = np.arange(24)

b1 = a.reshape(4, 6)
print(b1.shape, np.array_equal(a, b1.ravel()))

b2 = b1.reshape(2, 3, 4)
print(b2.shape, np.array_equal(a, b2.ravel()))

b3 = b2.reshape(2, 2, 6)
print(b3.shape, np.array_equal(a, b3.ravel()))

b4 = b3.reshape(24)
print(b4.shape, np.array_equal(a, b4.ravel()))
```

Метод reshape обычно не копирует данные, а лишь создаёт новое представление (view) с изменёнными метаданными о форме и шагах (strides), однако он вынужден создать копию в памяти, если элементы исходного массива расположены в ней не последовательно

```python
#Задание 6. Блок 2

import numpy as np

M = np.arange(12).reshape(3, 4)
f = M.flatten()
r = M.ravel()

print(f.base is None)
print(r.base is not None)

f[0] = 999
print(M)

r[1] = 888
print(M)
```

Метод ravel() стоит использовать для экономии памяти и скорости, когда нужно только читать данные или намеренно менять исходный массив, а flatten() когда требуется независимая копия, изменение которой гарантированно не испортит оригинал

```python
#Задача 7. Блок 2

import numpy as np

v = np.arange(5)

print(v[:, None].shape)
print(v[None, :].shape)
print(np.squeeze(v[:, None]).shape)

m = np.arange(6).reshape(2, 3).T
print(m.shape)
print(m)

res = v[:, None] + v[None, :]
print(res.shape)
print(res)
```

Операция v[:, None] превращает одномерный вектор в матрицу-столбец (5, 1), v[None, :] — в матрицу-строку (1, 5), np.squeeze удаляет лишнюю ось единичной длины, возвращая форму (5,), .T меняет местами строки и столбцы матрицы (получается форма (3, 2) со значениями [[0, 3], [1, 4], [2, 5]), а их сумма v[:, None] + v[None, :] за счёт механизма broadcasting создаёт матрицу формы (5, 5), которая представляет собой таблицу сложения всех элементов вектора v друг с другом

```python
#Задача 8. Блок 2

import numpy as np

A = np.arange(6).reshape(2, 3)
B = A.T

print(B.base is not None)

B[0, 1] = 999
print(A)
```

Транспонирование .T создаёт представление (view), а не копию, потому что NumPy не перемещает сами данные в памяти, а просто меняет местами значения в метаданных shape и strides (шаги чтения по осям), из-за чего изменение элемента в B напрямую меняет ту же ячейку памяти в A

```python
#Задача 9. Блок 3

import numpy as np

A = np.arange(6).reshape(2, 3)
B = np.arange(6).reshape(3, 2)

m1 = A @ B
m2 = A.T @ B.T
m3 = A * A

print(m1.shape)
print(m1)

print(m2.shape)
print(m2)

print(m3.shape)
print(m3)

D = np.diag([2, 3])
print(D * D)
print(D @ D)
```

Операция A * B выдаёт ошибку, потому что * выполняет поэлементное умножение и требует одинаковых (или совместимых для broadcasting) форм матриц, тогда как @ выполняет классическое матричное умножение (строка на столбец), а одинаковый результат обе операции дают, например, при умножении диагональной матрицы саму на себя (D * D == D @ D) или для нулевых матриц одинакового размера

```python
#Задача 10. Блок 3

import numpy as np

M = np.arange(12).reshape(3, 4)

print(M.sum(axis=0))
print(M.sum(axis=1))
print(M.mean(axis=0))
print(M.mean(axis=1))
print(M.argmax(axis=0))
print(M.argmax(axis=1))

small = np.array([[1, 2], [3, 4]])
print(small.sum(axis=0))
```

Операции с axis=0 сворачивает матрицу вертикально по строкам и выдают суммы, средние значения и индексы максимумов для каждого столбца (что подтверждает маленький пример, где 1+3=4 и 2+4=6), а операции с axis=1 сворачивают её горизонтально по столбцам, вычисляя те же показатели для каждой строки
