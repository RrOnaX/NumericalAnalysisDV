**Вывод:** Запись без экспоненциальной формы неоднозначна для **целых чисел без десятичной точки, которые оканчиваются на нули** (`1200`, `250000`). Без степени десятки невозможно понять, являются ли конечные нули результатом точного измерения или просто заполняют разряды до запятой.


```python
#Задача 7
x = 4.738291
delta = 0.02
x_rounded = round(x, 2)
print(f'Честная запись: {x_rounded} ± {delta}')
```

### Задача 7. Вывод
Погрешность $\Delta = 0.02$ имеет последнюю значащую цифру в разряде сотых, поэтому само число нужно округлить до того же разряда: **$4.74 \pm 0.02$**. В числе осталось **3 значащие цифры** (`4`, `7`, `4`).

**Вывод:** В записи $4.738291 \pm 0.02$ лишние знаки после сотых (`8291`) не имеют физического смысла, так как ошибка $\pm 0.02$ уже делает ненадёжной вторую цифру после запятой. Оставляя эти разряды, мы создаём ложную видимость высокой точности измерения.


```python
#Задача 8
import random

nums = [0.5, 1.5, 2.5, 3.5, -0.5, -1.5, -2.5]
print("Banker's rounding:", [round(n) for n in nums])
random.seed(42)
N = 1_000_000
arr = [random.randint(0, 9) + 0.5 for _ in range(N)]

sum_banker = sum(round(x) for x in arr)
sum_school = sum(int(x + 0.5) for x in arr)

shift = sum_school - sum_banker
print(f'Сумма (banker’s rounding): {sum_banker}')
print(f'Сумма (школьное правило):  {sum_school}')
print(f'Накопленный сдвиг:         {shift}')
```

### Задача 8. Вывод
Python использует округление до ближайшего чётного (*banker's rounding*): `0.5 -> 0`, `1.5 -> 2`, `2.5 -> 2`. Это сделано для того, чтобы половина пограничных значений округлялась вверх, а половина — вниз, не создавая перекоса. На массиве из $10^6$ чисел школьное правило «всегда вверх» накапливает систематический сдвиг около **$+500\,000$**, что в финансах критично, так как при миллионах операций приводит к появлению «лишних» денег из воздуха и нарушению баланса.


```python
#Задача 9
import math

terms = [1 / k for k in range(1, 101)]

s_exact = math.fsum(terms)

s_loop = 0.0
for t in terms:
    s_loop += t

s_round3 = sum(round(t, 3) for t in terms)

delta_a = abs(s_loop - s_exact)
rel_a = (delta_a / s_exact) * 100

delta_b = abs(s_round3 - s_exact)
rel_b = (delta_b / s_exact) * 100

print(f'Эталон (math.fsum):    {s_exact:.16f}')
print(f'(а) Без округления:    {s_loop:.16f} | ΔS = {delta_a:.2e} | δS = {rel_a:.2e}%')
print(f'(б) Округление до 3 зн: {s_round3:.16f} | ΔS = {delta_b:.6f} | δS = {rel_b:.4f}%')

if delta_a > 0:
    print(f'Ошибка (б) больше в {delta_b / delta_a:.2e} раз')
else:
    print('Ошибка (а) равна 0 (в пределах точности float), поэтому ошибка (б) больше в бесконечное число раз')
```

### Задача 9. Вывод
При вычислении без промежуточных округлений результат практически совпадает с эталоном `math.fsum` (абсолютная погрешность равна $0$ или порядка $10^{-16}$ в зависимости от способа суммирования). При округлении каждого слагаемого до 3 знаков абсолютная погрешность вырастает до **$\Delta S \approx 0.001378$** ($\delta S \approx 0.0266\%$), что более чем в **триллион раз ($> 10^{12}$)** превышает погрешность машинного представления.

**Правило:** Никогда не округляйте промежуточные результаты - проводите все вычисления с максимально возможной точностью и округляйте только финальный ответ.


```python
#Задача 10
import math

a = 0.1 + 0.2
b = 0.3

print(f'0.1 + 0.2 = {a:.17f}')
print(f'0.3       = {b:.17f}')
print('Сравнение через == :', a == b)
print('Сравнение через math.isclose:', math.isclose(a, b))

delta = abs(a - b)
rel_delta = (delta / abs(b)) * 100

print(f'Абсолютная погрешность Δ:    {delta:.2e}')
print(f'Относительная погрешность δ: {rel_delta:.2e}%')
```

### Задача 10. Вывод
* **Результаты расчётов:** `0.1 + 0.2` в памяти компьютера хранится как `0.30000000000000004`, поэтому прямое сравнение `== 0.3` возвращает `False`, а безопасное сравнение `math.isclose` - `True`. Абсолютная погрешность равна $\Delta \approx 5.55 \cdot 10^{-17}$, относительная - $\delta \approx 1.85 \cdot 10^{-14}\%$.

**Объяснение:**
В привычной нам десятичной системе дробь $1/3$ нельзя записать конечным числом цифр - получается бесконечная дробь $0.3333...$, которую приходится обрезать. Точно так же в двоичной системе, которой пользуется компьютер, обычное число $0.1$ превращается в бесконечную периодическую дробь. Поскольку память процессора ограничена, он обрезает этот бесконечный «хвост» при сохранении чисел $0.1$ и $0.2$, из-за чего при их сложении возникает крошечная погрешность округления.
