```python
import math

def relative_error(approx, exact):
    if exact == 0:
        return float('inf')
    return abs(approx - exact) / abs(exact)

pairs = [
    ('pi', 3.14, math.pi),
    ('e', 2.718, math.e),
    ('g', 9.8, 9.81),
    ('1/3', 0.333, 1 / 3),
]
for name, approx, exact in pairs:
    abs_err = abs(approx - exact)
    rel_err = relative_error(approx, exact)
    print(f'{name}: Δx = {abs_err:.6f}, δx = {rel_err * 100:.4f}%')

print('Zero test:', relative_error(0.5, 0))
```
